```python
# Generated by Google Gemini. Edited by Brian Hahne.
# Prompt user for a single-digit expression (e.g., 5+3)
user_input = input("Enter a single-digit calculation (e.g., 5+3): ").strip()
# Validate that the input is exactly 3 characters long
if len(user_input) != 3:
    print("Error: Input must be exactly 3 characters (e.g., 5+3).")
else:
    # Extract characters using string indexing
    char1 = user_input[0]
    operator = user_input[1]
    char2 = user_input[2]
    # Validate that the first and third characters are single digits
    if not (char1.isdigit() and char2.isdigit()):
        print("Error: Please ensure you enter single-digit numbers.")
    else:
        # Convert digit strings to integers
        num1 = int(char1)
        num2 = int(char2)
        # Initialize an error flag and result variable
        valid_op = True
        result = 0
        # Perform calculation based on the operator
        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            # Check for division by zero
            if num2 == 0:
                valid_op = False
                print("Error: Cannot divide by zero.")
            else:
                result = num1 / num2
        else:
            valid_op = False
            print(f"Error: '{operator}' is not a valid operator.")
        # Print the result if the operation succeeded
        if valid_op:
            print(f"The result is {result}")
```
